```python
# Cell 1: Environment Setup & Custom Path Routing
import sys
import os
import seaborn as sns
import matplotlib.pyplot as plt

# Route our environment paths so this notebook can import code from the /src folder
sys.path.append(os.path.abspath('../'))

from src.data_processing import load_and_clean_data
from src.analytical_core import calculate_cohort_matrix, calculate_rfm_segments
from src.operational_deep_dive import analyze_delivery_impact
```

```python
# Cell 2: Data Cleaning and Ingestion Influx
# This triggers our multi-table merge and cleans up datetimes
master_df = load_and_clean_data(data_dirpath='../data/')
```

```python
# Cell 3: Process and Plot the Customer Cohorts
# Build the cohort data matrix and visualize user drop-offs via a heatmap
retention = calculate_cohort_matrix(master_df)

plt.figure(figsize=(14, 8))
sns.heatmap(
    retention.iloc[:, 0:12], # Focus on the first year of activity
    annot=True, 
    fmt='.1%', 
    cmap='Blues', 
    vmin=0.0, 
    vmax=0.08 # Cap max scale at 8% so subtle month-over-month trends stand out
)
plt.title('Olist Marketplace Customer Cohort Retention (Months 0-12)')
plt.ylabel('Acquisition Cohort Month')
plt.xlabel('Months Active Since First Purchase')
plt.show()
```

```python
# Cell 4: Evaluate Segment Profiling Distributions
# Run the custom-binned behavioral RFM model
rfm_results = calculate_rfm_segments(master_df)

print("\n--- Customer Segment Distribution Breakdown ---")
print(rfm_results['Segment'].value_counts())
```

```python
# Cell 5: Operational Diagnostics Deep Dive
# Compute how delivery delays explicitly damage customer reviews
operational_impact = analyze_delivery_impact(master_df)

print("\n--- Logistics Performance Impact on Sentiment ---")
print(operational_impact)
```
